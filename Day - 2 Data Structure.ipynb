Q11. Find the Second Largest Number

```python
sales = [1200, 5000, 3200, 8000, 2500]
```

```python
largest = sales[0]
second_largest = float("-inf")

for x in sales:
    if x > largest:
        second_largest = largest
        largest = x

    elif x > second_largest and x != largest:
        second_largest = x
print(second_largest)
```
